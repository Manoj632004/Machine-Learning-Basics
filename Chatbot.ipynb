```python
from langchain import LlamaCpp
llm = LlamaCpp(
    model_path="Phi-3-mini-4k-instruct-fp16.gguf",
    n_gpu_layer=-1,
    max_tokens=500,
    n_ctx=2048,
    seed=42,
    verbose = False
)
```

```python
prompt = "The Code is not Running"
```

```python
output=llm(
    f"<|user|>\n{prompt}<\n|assistant|>",
    max_tokens=256,
    stop=["<|end|>"],
    echo=True
)
print(output)
```
